```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/raw/Train.csv")
df_test = pd.read_csv("../data/raw/Test.csv")
print(f"# Rows: {df.shape[0]}")
print(f"# Cols: {df.shape[1]}")
```

```python
target = df["is_climate_sensitive"]
cnts = target.value_counts()
plt.figure(figsize=(4,4))
plt.bar(cnts.index.astype(str), cnts.values, width=.75)
plt.title("Class Distribution of Target (0, 1)")
plt.ylabel("Counts")
plt.xlabel('Classes')
plt.show()
```
